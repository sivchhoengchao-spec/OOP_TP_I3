```python
class Greeting:
    """A class for greeting a user."""

    def __init__(self, name):
        self.name = name

    def say_hello(self):
        """Print a greeting message."""
        print(f"Hello, {self.name}!")


def main1():
    """Get the user's name and display a greeting."""
   
    name = input("Enter your name: ").strip()

    greeting = Greeting(name)
    greeting.say_hello()


if __name__ == "__main__":
    print("EX1:")
    main1()

        
```

```python
class Rectangle:
    """A rectangle with a length and a width."""

    def __init__ (self, length, width):
        if length <= 0 or width <= 0:
            raise ValueError("length and width must be greater than 0")
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

    def display(self):
        print(f"Length    : {self.length}")
        print(f"Width     : {self.width}")
        print(f"Area      : {self.area()}")
        print(f"Perimeter : {self.perimeter()}")


def read_number(message):
    """Ask until the user types a valid number."""
    while True:
        try:
            return float(input(message))
        except ValueError:
            print("Please enter a number, for example 5 or 3.5")


def main2():
    length = read_number("Enter the length of the rectangle: ")
    width = read_number("Enter the width of the rectangle: ")
    try:
        rect = Rectangle(length, width)
    except ValueError as error:
        print("Error:", error)
    else:
        rect.display()


if __name__ == "__main__":
    print("EX2:")
    main2()
```

```python
class HundredsCounter:
    """Counts how many complete hundreds are in a non-negative number."""
    def __init__(self, number):
        if number < 0:
            raise ValueError("Number must be non-negative.")

        self.number = number

    def count_hundreds(self):
        return self.number // 100


"""Ask the user for a number and print its complete hundreds."""
def main4():
    try:
        number = int(input("Enter a non-negative Number : "))
        counter = HundredsCounter(number)
        print("Complete Hundreds in the number:", counter.count_hundreds())
    except ValueError as e:
        print("Error : ", e)

if __name__ == "__main__":
    print("EX4:")
    main4()
```

```python
# Keeps the entered number and calculates the following integer.
class LuckyNumberGame:
    """Return the integer immediately after a user's number."""

    def __init__(self, number: int) :
        self.number: int = number  # Save the number for this game.

    def get_next_number(self) :
        """Calculate and return the next integer."""
        return self.number + 1  # Add one to get the next integer.


def main5():
    """Run rounds of the game until the user chooses to stop."""
    while True:  # Start another round unless the user chooses to stop.
        try:
            # Convert the user's input from text to an integer.
            number = int(input("Please enter an integer: "))
        except ValueError:
            # Handle text or other input that cannot be converted to an integer.
            print("Please enter a valid integer.")
            continue  # Skip the rest of this round and ask for a number again.

        # Show the integer that comes after the user's number.
        game = LuckyNumberGame(number)
        print(f"The next number is: {game.get_next_number()}")

        # Repeat the prompt until the user enters "yes" or "no".
        while True:
            answer = input("Do you want to play again? (Yes/No): ").strip().lower()
            if answer in {"yes", "no"}:
                break
            print("Please enter Yes or No.")

        if answer == "no":
            break  # Exit the game loop when the user is finished.


# Run the game only when this file is started directly.
if __name__ == "__main__":
    print("EX5:")
    main5()
```
